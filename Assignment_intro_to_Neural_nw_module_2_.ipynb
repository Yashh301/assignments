# **Objective :  The primary objective is to predict the output value y based on two input features x1 and x2 using a basic neural network model with the equation y=3x1+4x2. The neural network will be trained using a feed-forward and back-propagation mechanism to minimize the error between the predicted and actual values.**

Importing the necessary libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

Defining some popular activation function used in neural networks, however not required in this problem.

```python
def sigmoid(x): # outputs between 0 and 1.
    return 1 / (1 + np.exp(-x))

def tanh(x): # Gives the output between -1 to 1.
    return np.tanh(x)


def relu(x): # gives the output zero (0) till there is a positive non zero value in the input range(0 - Infinity).
    return np.maximum(0, x)
```

```python
a= sigmoid(-2)
print(a)
b= sigmoid(0.2)
print(b)
c = sigmoid(0)
print(c)
d=sigmoid(6)
print(d)
```

```python
a= tanh(-2)
print(a)
b= tanh(0.2)
print(b)
c = tanh(0)
print(c)
d=tanh(3)
print(d)
```

```python
a= relu(-2)
print(a)
b= relu(0.2)
print(b)
c = relu(0)
print(c)
d=relu(3)
print(d)
```

Defining training data.

Hardcoding each and every step instead of using the framworks like tensorlfow.

## **Feed Forward.**

```python
#Defining Training Datax1, x2 = 3, 2  # Input features
y_actual = 17   # Target output

#Initializing Weights Randomly
np.random.seed(42)  # For reproducibility
w1, w2 = np.random.uniform(1, 10, 2)  # Random weights between 1 and 10
learning_rate = 0.01

epochs = 20
errors = []  # To track error over epochs
ypred_values = []  # To track predictions

#Training iteration
for epoch in range(epochs):
    # Forward Pass (Prediction)
    y_pred = x1 * w1 + x2 * w2

```

## **Training, Error calculation and back propagation.**

```python
epochs = 20
errors = []  # To track error over epochs
ypred_values = []  # To track predictions

#Training iteration
for epoch in range(epochs):
    # Forward Pass (Prediction)
    y_pred = x1 * w1 + x2 * w2

# Error Calculation
    error = (y_actual - y_pred) ** 2
    errors.append(error)
    ypred_values.append(y_pred)

# Back Propagation - Compute Gradients (Partial Derivatives of Error w.r.t w1 and w2)
    dE_dw1 = -2 * (y_actual - y_pred) * x1  # Gradient for w1
    dE_dw2 = -2 * (y_actual - y_pred) * x2  # Gradient for w2

# Update Weights using Gradient Descent
    w1 -= learning_rate * dE_dw1
    w2 -= learning_rate * dE_dw2

# Print progress
    print(f"Epoch {epoch+1}: y_pred = {y_pred:.4f}, Error = {error:.4f}")


print(f"Final weights: w1 = {w1:.4f}, w2 = {w2:.4f}")
print(f"Final Prediction: y_pred = {ypred_values[-1]:.4f}, Expected: {y_actual}")
```

## **Visualization.**

```python
# Step 5: Visualization
plt.figure(figsize=(10, 4))

# Error Plot
plt.subplot(1, 2, 1)
plt.plot(range(epochs), errors, marker='o', linestyle='-', color='r')
plt.xlabel('Epochs')
plt.ylabel('Error')
plt.title('Error Reduction Over Time')

# y_pred Plot
plt.subplot(1, 2, 2)
plt.plot(range(epochs), ypred_values, marker='o', linestyle='-', color='b')
plt.xlabel('Epochs')
plt.ylabel('Predicted y')
plt.title('Predicted y Over Time')

plt.show()
```
