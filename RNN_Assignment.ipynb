```python
# Importing the libraries.
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python

# Loading IMDB dataset while keeping the top 10,000 words but skipping the top 20 most common words
num_words = 10000 #Keepin only top 10000 words.
skip_top = 20  # Ignore top 20 most common words

(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=num_words, skip_top=skip_top)

# Defining max length of sequence for padding
maxlen = 200

# Pad sequences to ensure all reviews are the same length
x_train = pad_sequences(x_train, maxlen=maxlen)
x_test = pad_sequences(x_test, maxlen=maxlen)
```

```python
#Defining the model.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.layers import Dropout
# Define model
model = Sequential([
    Embedding(input_dim=num_words, output_dim=32, input_length=maxlen),  # Ensure input_length is set
    Dropout(0.5),
    SimpleRNN(32, return_sequences=False),  # SimpleRNN layer with 32 units
    Dense(1, activation='sigmoid')  # Output layer for binary classification
])

# Building the model explicitly
model.build(input_shape=(None, maxlen))

# Compiling the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Printing model summary
model.summary()
```

```python
# Training the model
history = model.fit(
    x_train, y_train,
    batch_size=128,
    epochs=10,
    validation_split=0.2
)
```

```python
# Testing the accuracy
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print(f"Test Accuracy: {test_accuracy:.4f}")
print(f"Test Loss: {test_loss:.4f}")
```
